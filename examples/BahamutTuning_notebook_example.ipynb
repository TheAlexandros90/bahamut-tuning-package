# Bahamut Tuning notebook example

This notebook demonstrates the planning stage, code generation, and optional UI for the Bahamut hyperparameter-tuning subsystem.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'src').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

SRC_DIR = PROJECT_ROOT / 'src'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from sklearn.datasets import load_diabetes
from bahamut_tuning import HyperTuneWorkbench, TuningSelection
```

```python
data = load_diabetes(as_frame=True)
df = data.frame.copy()
df['target'] = data.target

tuner = HyperTuneWorkbench(
    df=df,
    target=['target'],
    task_type='regression',
    selected_algorithm='RandomForestRegressor',
    recommended_algorithms=['RandomForestRegressor', 'ElasticNet'],
    scorer='neg_root_mean_squared_error',
    metadata={'tuning_policy': 'random'},
)
```

```python
plan = tuner.plan(TuningSelection(tuning_strategy='auto', iterations=25))
plan.to_dict()
```

```python
print(tuner.generate_code(plan))
```
